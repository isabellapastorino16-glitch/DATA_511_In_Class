# Exercise 2 (5 points)

Define the term objective function in the context of linear programming and explain its role in optimization.


The **objective function** is a linear function repesenting the goal of the optimization, such as maximizing profit or minimizing cost. It is expressed in terms of decision variables and is subject to contraints. The linear programming process finds the values of these variables that optimize this function while satisfying all given constraints.

# Exercise 4 (10 points)

A pastry shop produces two types of pastries: Croissants (X) and Danishes (Y). Each pastry requires:

| Resource | Croissant (X) | Danish (Y) | Available | 
| --- | --- | --- | --- |
| Flour (grams) | 200 | 100 | 5000 |
| Labor (minutes) | 10 | 15 | 400 |
| Profit per unit | $2 | $3 | |

The owner wants to maximize profit while not exceeding available resources. Formulate and solve this as a linear programming problem using Python.

```python
# we want to maximize 2X + 3Y
# subject to constraints:
# 200X + 100Y <= 5000 (flour constraint)
# 10X + 15Y <= 400 (labor constraint)
# X >= 0 (non-negativity constraint)
# Y >= 0 (non-negativity constraint)

from scipy.optimize import linprog

# Coefficients of the objective function (we want to maximize 2X + 3Y)
c = [-2, -3]  # linprog does minimization, so we use negative coefficients

# Coefficients of the inequality constraints
A = [[200, 100],  # flour constraint
     [10, 15]]    # labor constraint
b = [5000, 400]

# Bounds for X and Y
x_bounds = (0, None)
y_bounds = (0, None)

# Solve the linear programming problem
res = linprog(c, A_ub=A, b_ub=b, bounds=[x_bounds, y_bounds])

#display the results
if res.success:
    print("Optimal solution found:")
    print("X (crossaint) =", res.x[0])
    print("Y (danish)=", res.x[1])
    print("Maximum value of objective function:", -res.fun)
else:
    print("No optimal solution found.")
```

# Exercise 5 (10 points)

Consider the context presented in Exercise 4. This time solve the program as an interger linear programming. Formulate and solve the problem as an integer programming problem in Python.

```python
import numpy as np
from scipy.optimize import milp, LinearConstraint, Bounds

#objective function coefficients (maximize 2X + 3Y)
c = [-2, -3]

#inequality constraints (flour and labor)
A = np.array([
    [200, 100],  # flour constraint
    [10, 15]     # labor constraint
])
b = np.array([5000, 400])

# setting the constraints
linear_constraint = LinearConstraint(A, -np.inf, ub = b)

# setting the bounds for the decision variables (X and Y >= 0)
bounds = Bounds([0, 0], [np.inf, np.inf])

#integer constraints for the decision variables (if X and Y must be integers)
integrality = np.array([1, 1])  # 1 indicates the variable must be an integer

# solve the integer linear programming problem 
result = milp(c=c, constraints=[linear_constraint], bounds=bounds, integrality=integrality)

#display the result
if result.success:
    print("Optimal solution found:")
    print("X =", result.x[0])
    print("Y =", result.x[1])
    print("Objective value =", -result.fun)  # negate because we minimized the negative
else:
    print("No optimal solution found.")
```
