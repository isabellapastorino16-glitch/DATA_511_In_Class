# Exercise 4 (10 points)

You are given three development scenarios for a hotel project with associated prior probabilities and annual earnings over a 10-year horizon. The discount rate is 5%. Your task is to compute the expected present value of the project using the formula for an ordinary annuity. The data is shown below:

```
scenarios = ["Low", "Medium", "High"]
probabilities = [0.2, 0.5, 0.3]
earnings = [1.1, 1.6, 2.0]  # in billions
discount_rate = 0.05
term = 10
```

Write a Python function to compute the present value of an ordinary annuity and use it to calculate the expected present value of the project based on the scenarios above.

````python
scenarios = ["Low", "Medium", "High"]
probabilities = [0.2, 0.5, 0.3]
earnings = [1.1, 1.6, 2.0]  # in billions
discount_rate = 0.05
term = 10

#calculate the formula to compute the present value of earnings
def present_value_earnings(earnings,r, t): #r for discount rate, t for term
    return earnings / r * (1 - (1 + r) ** -t )

#compute the present value of earnings for each scenario
present_values = [present_value_earnings(e, discount_rate, term) for e in earnings]

#compute the expected present value of earnings
expected_present_value = sum(p * pv for p, pv in zip(probabilities, present_values))

print("Expected Present Value of Earnings:", expected_present_value)
#print the present values for each scenario
for scenario, pv in zip(scenarios, present_values):
    print(f"Present Value for {scenario} Scenario: {pv}")
````
