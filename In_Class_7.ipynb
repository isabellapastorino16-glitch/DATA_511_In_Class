# Exercise 3 (5 points)

Write a Python function to simulate tossing a coin 10,000 times. Use 0 for tails and 1 for heads.

```python
import numpy as np

#simulate a sedries of coin flips with 0 as tails and 1=heads

n_sim = 10000 # number of simulations
results = [np.random.choice([0, 1]) for _ in range(n_sim)]

#calculate the proportion of heads
prop_heads = np.mean(results)
print("Proportion of heads:", prop_heads)
```

Using Python, simulate 1000 regressions and store coefficients of each run.

```python
import pandas as pd
import statsmodels.formula.api as smf

import matplotlib.pyplot as plt

#storing betas
betas = []
for i in range(1000):
    #simulate some data
    np.random.seed(i)  # set seed for reproducibility
    x = np.random.normal(.6, .95, size = 120)
    y = 50 + 20*x + np.random.normal(0, 2, size=120)

    df = pd.DataFrame({'x': x, 'y': y})

    model = smf.ols('y ~ x', data=df).fit()
    betas.append(model.params['x'])

#plotting the distribution of betas
plt.hist(betas, bins=30)
plt.xlabel('Beta Estimates')
plt.ylabel('Frequency')
plt.title('Distribution of Beta Estimates from 1000 Simulations')
plt.vlines(x=20, ymin=0, ymax=plt.gca().get_ylim()[1], color='r', linestyle='dashed', label='True Coefficient', linewidth=2)
plt.grid()
plt.show()
    
```
