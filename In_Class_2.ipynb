# Exercise 2 (4 points)

Briefly describe the difference between stochastic and non-stochastic prescriptive methods.

Non-stochastic methods do not directly incorporate randomness and focus on deterministic processes (ex. mathematical programming). Stochastic methods explicitly account for random variances, such as Monte Carlo simulations modeling.

# Exercise 4 (10 points)

Consider the `train.csv` data file. This data file contains information related to music tracks. The goal is to build a machine learning model to predict `BeatsPerMinute`. Explore the data.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('train.csv')
df.head()
```

```python
plt.hist(df['BeatsPerMinute'], bins=20, edgecolor='black', color='skyblue')
plt.ylabel('Frequency')
plt.title('Distribution of Beats Per Minute')
plt.grid()
plt.show()
```

The 'BeatsPerMinute' distribution looks symmetrical and normally distributed.

```python
plt.plot(df['Energy'], df['BeatsPerMinute'], 'o', color='orange')
plt.xlabel('Energy')
plt.ylabel('Beats Per Minute')
plt.title('Energy vs Beats Per Minute')
plt.grid()
plt.show()
```
